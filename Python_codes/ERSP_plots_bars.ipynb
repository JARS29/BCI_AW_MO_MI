```python
import pandas as pd
import numpy as np
from scipy.stats import wilcoxon 
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import os
from fooof import FOOOF
import numpy as np
from statsmodels.stats.multitest import multipletests
```

```python
root_data = 'CSV_ERSP_CH/' 
files = os.listdir(root_data)

tasks = ['Arm', 'Leg']
conds = ['AW', 'MI', 'MO']
CHs = ['C3', 'Cz', 'C4', 'PO7','Pz', 'PO8']

file_names = []
for ch in CHs:
    for cond in conds:
            for task in tasks:
                file_names.append('_'.join([ch, cond, task]))
file_names
```

```python
data_files = {}

for name in file_names:
    temp_files =[]
    for file in files:
        if name in file:
            temp_files.append(pd.read_csv(os.path.join(root_data, file), header=None))
    data_files[name] = pd.concat(temp_files)
data_files
```

```python
import numpy as np
from statsmodels.stats.multitest import multipletests

p_AW_MI = {}
p_AW_MO = {}

p_AW_ARM_LEG = {}
#false discovery rate (FDR)

# Collect all p-values for FDR correction
all_p_values = []

for ch in CHs:
    for task in tasks:
        p_AW_MI[ch+'_'+task]=[]
        p_AW_MO[ch+'_'+task]=[]
        if task=='Arm':
            p_AW_ARM_LEG[ch+'_Arm_Leg']=[]

        for i in range(len(data_files['C3_AW_Arm'].columns)):
            p_AW_MI[ch+'_'+task].append(wilcoxon(data_files[ch+'_AW_'+task].iloc[:, i], data_files[ch+'_MI_'+task].iloc[:, i], alternative='two-sided')[1])
            p_AW_MO[ch+'_'+task].append(wilcoxon(data_files[ch+'_AW_'+task].iloc[:, i], data_files[ch+'_MO_'+task].iloc[:, i], alternative='two-sided')[1])
            all_p_values.append(p_AW_MI[ch+'_'+task][-1])  # Add p-value to the list
            all_p_values.append(p_AW_MO[ch+'_'+task][-1])

            if task=='Arm':
                p_AW_ARM_LEG[ch+'_Arm_Leg'].append(wilcoxon(data_files[ch+'_AW_Leg'].iloc[:, i], data_files[ch+'_AW_'+task].iloc[:, i], alternative='two-sided')[1])
                all_p_values.append(p_AW_ARM_LEG[ch+'_Arm_Leg'][-1])

# Apply FDR correction (Benjamini-Hochberg procedure)
_, q_values, _, _ = multipletests(all_p_values, method='fdr_bh')

# Map corrected q-values back to the original p-value structures
q_value_idx = 0
for ch in CHs:
    for task in tasks:
        for i in range(len(p_AW_MI[ch + '_' + task])):  # Iterate over the actual p-values in the list
            p_AW_MI[ch + '_' + task][i] = q_values[q_value_idx] 
            q_value_idx += 1
        for i in range(len(p_AW_MO[ch + '_' + task])):
            p_AW_MO[ch + '_' + task][i] = q_values[q_value_idx]
            q_value_idx += 1

        if task == 'Arm':
            for i in range(len(p_AW_ARM_LEG[ch + '_Arm_Leg'])):
                p_AW_ARM_LEG[ch + '_Arm_Leg'][i] = q_values[q_value_idx]
                q_value_idx += 1
```

```python
import numpy as np
import matplotlib.pyplot as plt

# === Step 1: Load the ERSP matrix ===
filename = 'CSV_ERSP_CH/PO7_AW_Arm_S_1.csv'
ersp = np.loadtxt(filename, delimiter=',')

# === Step 2: Define time and frequency axes (match your MATLAB config) ===
times = np.linspace(-1000, 1500, ersp.shape[1])   # 4000 ms window, based on EEG.pnts
freqs = np.linspace(4, 40, ersp.shape[0])         # From 'freqs', 4–40 Hz over 80 steps

# === Step 3: Select frequency band (e.g., mu = 9–13 Hz) ===
mu_band = (freqs >= 9) & (freqs <= 13)
mu_power = ersp[mu_band, :]        # Select only mu-band rows
mu_avg_power = mu_power.mean(axis=0)  # Average across frequencies
beta_band = (freqs >= 18) & (freqs <= 24)
beta_power = ersp[beta_band, :]
beta_avg_power = beta_power.mean(axis=0)

# === Step 4: Plot power vs time ===
plt.figure(figsize=(10, 4))
plt.plot(times, beta_avg_power, color='purple', linewidth=2)
plt.axvline(0, color='black', linestyle='--', label='Go cue')  # Optional event marker

plt.title('Mu Band Power (9–13 Hz) over Time')
plt.xlabel('Time (ms)')
plt.ylabel('ERSP Power (dB)')  # If your MATLAB data was in dB
plt.grid(True)
plt.tight_layout()
plt.show()
```
