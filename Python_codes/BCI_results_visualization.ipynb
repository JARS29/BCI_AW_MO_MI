```python
df=offline_classification[offline_classification['Feature_Extraction']=='Combined'].drop(columns=['Feature_Extraction','Selected_Features'])
df_grouped = df.iloc[:,1:].groupby('Condition').agg(['mean', 'std'])
print(df_grouped)

# Calculate means and stds for error bars
means = df.groupby('Condition')['Accuracy'].mean()
stds = df.groupby('Condition')['Accuracy'].std()

# Create Plotly grouped bar chart with error bars
fig_combined = go.Figure()

for condition in df['Condition'].unique():
    cond_df = df[df['Condition'] == condition]
    fig_combined.add_trace(go.Bar(
        x=cond_df['Subject'],
        y=cond_df['Accuracy'],
        name=condition,
        marker_color=colors[condition],
        error_y=dict(
            type='data',
            array=[stds[condition]] * len(cond_df),  # same std for all bars in this condition
            visible=True,
            color ='rgba(0,0,0,0.5)'

        )
    ))

# Add chance level line
fig_combined.add_hline(y=0.5, line_width=1, line_color="gray")

# Update layout to show subject IDs on x-axis ticks
fig_combined.update_layout(
    barmode='group',
    xaxis_title='Subject',
    yaxis_title='Accuracy',
    title='Accuracy per Subject and Condition  - Combined FBCSP and ERP',
    legend_title_text='Condition',
    xaxis=dict(
        tickmode='array',
        tickvals=offline_classification['Subject'].unique(),
        ticktext=offline_classification['Subject'].unique()
    ),
    yaxis=dict(range=[0.3, 1.0], tick0=0.4, dtick=0.1),
)

fig_combined.show()
```

```python
fig_AW_seen = go.Figure()
fig_AW_new = go.Figure()
fig_MI = go.Figure()
fig_MO = go.Figure()


x_times = np.linspace(0,2.5,11)*1000  # Convert to milliseconds
all_online_AW_results_seen=online_ov_classification[online_ov_classification['Condition']=='AW_Seen']
all_online_AW_results_new=online_ov_classification[online_ov_classification['Condition']=='AW_New']
all_online_MI_results=online_ov_classification[online_ov_classification['Condition']=='MI']
all_online_MO_results=online_ov_classification[online_ov_classification['Condition']=='MO']

# Seen AW
plot_data_aw_seen = {}
plot_data_aw_new = {}
plot_data_MI = {}
plot_data_MO = {}

for subject in all_online_AW_results_seen.Subject.unique():

    fbcsp_mean = np.mean(all_online_AW_results_seen[all_online_AW_results_seen['Subject']==subject]['FBCSP_acc'])
    erp_mean = np.mean(all_online_AW_results_seen[all_online_AW_results_seen['Subject']==subject]['ERP_acc'])
    combined_mean = np.mean(all_online_AW_results_seen[all_online_AW_results_seen['Subject']==subject]['Combined_acc'])
    means = {'FBCSP': fbcsp_mean, 'ERP': erp_mean, 'Combined': combined_mean}
    best_model_name = max(means, key=means.get)
    plot_data_aw_seen[subject] = {
        'model_name': best_model_name,
        'accuracies': all_online_AW_results_seen[all_online_AW_results_seen['Subject']==subject][f'{best_model_name}_acc'].tolist()
    }


    fbcsp_mean = np.mean(all_online_AW_results_new[all_online_AW_results_new['Subject']==subject]['FBCSP_acc'])
    erp_mean = np.mean(all_online_AW_results_new[all_online_AW_results_new['Subject']==subject]['ERP_acc'])
    combined_mean = np.mean(all_online_AW_results_new[all_online_AW_results_new['Subject']==subject]['Combined_acc'])
    means = {'FBCSP': fbcsp_mean, 'ERP': erp_mean, 'Combined': combined_mean}
    best_model_name = max(means, key=means.get)
    plot_data_aw_new[subject] = {
        'model_name': best_model_name,
        'accuracies': all_online_AW_results_new[all_online_AW_results_new['Subject']==subject][f'{best_model_name}_acc'].tolist()
    }

    fbcsp_mean = np.mean(all_online_MI_results[all_online_MI_results['Subject']==subject]['FBCSP_acc'])
    erp_mean = np.mean(all_online_MI_results[all_online_MI_results['Subject']==subject]['ERP_acc'])
    combined_mean = np.mean(all_online_MI_results[all_online_MI_results['Subject']==subject]['Combined_acc'])
    means = {'FBCSP': fbcsp_mean, 'ERP': erp_mean, 'Combined': combined_mean}
    best_model_name = max(means, key=means.get)
    plot_data_MI[subject] = {
        'model_name': best_model_name,
        'accuracies': all_online_MI_results[all_online_MI_results['Subject']==subject][f'{best_model_name}_acc'].tolist()
    }

    fbcsp_mean = np.mean(all_online_MO_results[all_online_MO_results['Subject']==subject]['FBCSP_acc'])
    erp_mean = np.mean(all_online_MO_results[all_online_MO_results['Subject']==subject]['ERP_acc'])
    combined_mean = np.mean(all_online_MO_results[all_online_MO_results['Subject']==subject]['Combined_acc'])
    means = {'FBCSP': fbcsp_mean, 'ERP': erp_mean, 'Combined': combined_mean}
    best_model_name = max(means, key=means.get)
    plot_data_MO[subject] = {
        'model_name': best_model_name,
        'accuracies': all_online_MO_results[all_online_MO_results['Subject']==subject][f'{best_model_name}_acc'].tolist()
    }
# Sort subjects numerically
sorted_subjects = sorted(plot_data_aw_seen.keys(), key=lambda x: int(x))
y_subjects = [f"{subject}: {plot_data_aw_seen[subject]['model_name']}" for subject in sorted_subjects]
z_accuracies = [plot_data_aw_seen[subject]['accuracies'] for subject in sorted_subjects]
fig_AW_seen.add_trace(go.Heatmap(
    x=x_times,
    y=y_subjects,
    z=z_accuracies,
    text=z_accuracies,
    texttemplate="%{z:.2f}",
    colorscale=[[0.0, 'red'], [0.5, 'white'], [1.0, 'blue']],
    zmin=0,
    zmax=1,
    colorbar=dict(title="Accuracy"),
))

sorted_subjects = sorted(plot_data_aw_new.keys(), key=lambda x: int(x))
y_subjects = [f"{subject}: {plot_data_aw_new[subject]['model_name']}" for subject in sorted_subjects]
z_accuracies = [plot_data_aw_new[subject]['accuracies'] for subject in sorted_subjects]
fig_AW_new.add_trace(go.Heatmap(
    x=x_times,
    y=y_subjects,
    z=z_accuracies,
    text=z_accuracies,
    texttemplate="%{z:.2f}",
    colorscale=[[0.0, 'red'], [0.5, 'white'], [1.0, 'blue']],
    zmin=0,
    zmax=1,
    colorbar=dict(title="Accuracy"),
))

sorted_subjects = sorted(plot_data_MI.keys(), key=lambda x: int(x))
y_subjects = [f"{subject}: {plot_data_MI[subject]['model_name']}" for subject in sorted_subjects]
z_accuracies = [plot_data_MI[subject]['accuracies'] for subject in sorted_subjects]
fig_MI.add_trace(go.Heatmap(
    x=x_times,
    y=y_subjects,
    z=z_accuracies,
    text=z_accuracies,
    texttemplate="%{z:.2f}",
    colorscale=[[0.0, 'red'], [0.5, 'white'], [1.0, 'blue']],
    zmin=0,
    zmax=1,
    colorbar=dict(title="Accuracy"),
))


sorted_subjects = sorted(plot_data_MO.keys(), key=lambda x: int(x))
y_subjects = [f"{subject}: {plot_data_MO[subject]['model_name']}" for subject in sorted_subjects]
z_accuracies = [plot_data_MO[subject]['accuracies'] for subject in sorted_subjects]
fig_MO.add_trace(go.Heatmap(
    x=x_times,
    y=y_subjects,
    z=z_accuracies,
    text=z_accuracies,
    texttemplate="%{z:.2f}",
    colorscale=[[0.0, 'red'], [0.5, 'white'], [1.0, 'blue']],
    zmin=0,
    zmax=1,
    colorbar=dict(title="Accuracy"),
))

print('done')
```

```python
def plot_data_to_df(plot_data, condition):
    rows = []
    for subject, info in plot_data.items():
        accs = np.array(info['accuracies'])
        rows.append({
            'Subject': subject,
            'Condition': condition,
            'Model': info['model_name'],
            'Mean_Accuracy': np.mean(accs),
            'Std_Accuracy': np.std(accs)
        })
    return pd.DataFrame(rows)

df_aw_seen = plot_data_to_df(plot_data_aw_seen, 'AW_Seen')
df_aw_new = plot_data_to_df(plot_data_aw_new, 'AW_New')
df_mi = plot_data_to_df(plot_data_MI, 'MI')
df_mo = plot_data_to_df(plot_data_MO, 'MO')

df_unique = pd.concat([df_aw_seen, df_aw_new, df_mi, df_mo], ignore_index=True)
df_unique.groupby(['Condition']).agg(['mean']).drop(columns=['Subject'])
```
